```python
import pandas as pd
import numpy as np
import random
from datetime import datetime, timedelta

# 1. Setup our realistic parameters
hubs = ['Hyderabad', 'Mumbai', 'Delhi', 'Chennai', 'Bangalore']
carriers = ['BlueDart', 'FedEx', 'Delhivery', 'DHL', 'Shadowfax']
products = ['Insulin Glargine', 'COVID-19 mRNA Vaccine', 'Amoxicillin', 'Humira (Adalimumab)']

# 2. Generate Base Data
np.random.seed(42)
num_records = 5000

data = {
    'Shipment_ID': [f"SHP-{random.randint(10000, 99999)}" for _ in range(num_records)],
    'Product': np.random.choice(products, num_records),
    'Origin': np.random.choice(hubs, num_records),
    'Destination': np.random.choice(hubs, num_records),
    'Carrier': np.random.choice(carriers, num_records),
    'Spoiled_Flag': np.random.choice([0, 1], num_records, p=[0.94, 0.06]) # ~6% spoilage rate
}

df = pd.DataFrame(data)

# 3. Add Dates (and intentionally break some formats)
start_date = datetime(2025, 1, 1)
departure_dates = [start_date + timedelta(days=random.randint(0, 365)) for _ in range(num_records)]
arrival_dates = [d + timedelta(hours=random.randint(12, 72)) for d in departure_dates]

# Intentionally create "dirty" date strings
df['Departure_Time'] = [d.strftime('%Y-%m-%d %H:%M:%S') if random.random() > 0.2 else d.strftime('%d/%m/%Y %H:%M') for d in departure_dates]
df['Arrival_Time'] = [d.strftime('%Y-%m-%d %H:%M:%S') for d in arrival_dates]

# 4. Inject "Enterprise Dirt" (Nulls, Typos, Whitespace)
# Make 5% of Carriers NULL
df.loc[df.sample(frac=0.05).index, 'Carrier'] = np.nan

# Add trailing spaces and lowercase typos to Destinations
df['Destination'] = df['Destination'].apply(lambda x: x + " " if random.random() > 0.8 else x)
df['Destination'] = df['Destination'].apply(lambda x: x.lower() if random.random() > 0.9 else x)

# 5. Export
df.to_csv('dirty_pharma_shipments.csv', index=False)
print("Dataset generated successfully: 'dirty_pharma_shipments.csv'")
print("Rows:", len(df))
```

```python
import requests
import pandas as pd
import time

# 1. Define our Logistics Hubs and their coordinates (Lat, Lon)
hubs = {
    'Hyderabad': (17.3850, 78.4867),
    'Mumbai': (19.0760, 72.8777),
    'Delhi': (28.7041, 77.1025),
    'Chennai': (13.0827, 80.2707),
    'Bangalore': (12.9716, 77.5946)
}

# 2. Setup Open-Meteo API parameters (Historical Data)
start_date = '2025-01-01'
end_date = '2025-06-30' # First half of the year
base_url = "https://archive-api.open-meteo.com/v1/archive"

all_weather_data = []

print("Fetching historical weather data. This will take a few seconds...")

# 3. Loop through each hub and pull data
for city, coords in hubs.items():
    params = {
        "latitude": coords[0],
        "longitude": coords[1],
        "start_date": start_date,
        "end_date": end_date,
        "daily": "temperature_2m_max",
        "timezone": "Asia/Kolkata"
    }
    
    response = requests.get(base_url, params=params)
    
    if response.status_code == 200:
        data = response.json()
        
        # Open-Meteo returns data in lists. Let's zip them into rows.
        dates = data['daily']['time']
        max_temps = data['daily']['temperature_2m_max']
        
        for date, temp in zip(dates, max_temps):
            all_weather_data.append({
                'Hub_City': city,
                'Date': date,
                'Max_Temp_C': temp
            })
        print(f"✅ Downloaded data for {city}")
    else:
        print(f"❌ Failed to get data for {city}. Status: {response.status_code}")
    
    # Be polite to the API, wait half a second between requests
    time.sleep(0.5)

# 4. Save to CSV
weather_df = pd.DataFrame(all_weather_data)
weather_df.to_csv('weather_data.csv', index=False)
print("\nWeather data successfully saved to 'weather_data.csv'")
```

```python
# ============================================================
# SPOILAGE THRESHOLD HYPOTHESIS TEST (Chi-Square)
# ============================================================
# Business Rule Under Test:
#   Shipments departing when Origin Temp > 30°C AND Transit > 40hr
#   have an elevated spoilage rate vs. all other shipments.
#
# This test runs ONLY on the ~49% of shipments that successfully
# joined to a weather record (i.e., have a non-null temperature).
# ============================================================
import pandas as pd
import numpy as np
from scipy.stats import chi2_contingency
from datetime import datetime

# Load the shipment data
shipments = pd.read_csv('../data/raw/dirty_pharma_shipments.csv')
weather = pd.read_csv('../data/raw/jan_2025_to_july_2025_dataset.csv')

# Parse dates (mirror the Snowflake CLEAN layer logic)
def parse_departure(val):
    for fmt in ['%Y-%m-%d %H:%M:%S', '%d/%m/%Y %H:%M']:
        try:
            return datetime.strptime(str(val).strip(), fmt)
        except ValueError:
            continue
    return pd.NaT

shipments['Departure_Parsed'] = shipments['Departure_Time'].apply(parse_departure)
shipments['Arrival_Parsed'] = pd.to_datetime(shipments['Arrival_Time'], errors='coerce')

# Compute transit time in hours
shipments['Transit_Hours'] = (shipments['Arrival_Parsed'] - shipments['Departure_Parsed']).dt.total_seconds() / 3600

# Prepare join keys (mirror Snowflake INITCAP(TRIM(...)))
shipments['Origin_Clean'] = shipments['Origin'].str.strip().str.title()
shipments['Departure_Date'] = shipments['Departure_Parsed'].dt.date
weather['Date_Parsed'] = pd.to_datetime(weather['Date']).dt.date

# LEFT JOIN shipments to weather on (Origin Hub, Departure Date)
merged = shipments.merge(
    weather,
    left_on=['Origin_Clean', 'Departure_Date'],
    right_on=['Hub_City', 'Date_Parsed'],
    how='left'
)

# Filter to matched-only population (the ~49% with weather data)
matched = merged.dropna(subset=['Max_Temp_C', 'Transit_Hours'])

# Apply excursion threshold: Temp > 30°C AND Transit > 40 hours
matched['is_excursion'] = (matched['Max_Temp_C'] > 30) & (matched['Transit_Hours'] > 40)

# Build contingency table
excursion_spoiled = matched[(matched['is_excursion']) & (matched['Spoiled_Flag'] == 1)].shape[0]
excursion_ok = matched[(matched['is_excursion']) & (matched['Spoiled_Flag'] == 0)].shape[0]
control_spoiled = matched[(~matched['is_excursion']) & (matched['Spoiled_Flag'] == 1)].shape[0]
control_ok = matched[(~matched['is_excursion']) & (matched['Spoiled_Flag'] == 0)].shape[0]

contingency = np.array([[control_ok, control_spoiled],
                         [excursion_ok, excursion_spoiled]])

chi2, p_value, dof, expected = chi2_contingency(contingency)

exc_rate = excursion_spoiled / (excursion_spoiled + excursion_ok) * 100
ctrl_rate = control_spoiled / (control_spoiled + control_ok) * 100

print('=================================================================')
print('   SPOILAGE THRESHOLD HYPOTHESIS: CHI-SQUARE TEST')
print('=================================================================')
print()
print('Business Rule Under Test:')
print('  Excursion group: Origin Temp > 30°C AND Transit > 40 Hours')
print()
print('Contingency Table:')
print(f'                 Not Spoiled  Spoiled')
print(f'Control Group    {control_ok:>10}  {control_spoiled:>7}')
print(f'Excursion Group  {excursion_ok:>10}  {excursion_spoiled:>7}')
print()
print('Results:')
print(f'  Excursion spoilage rate: {exc_rate:.2f}%')
print(f'  Control spoilage rate:   {ctrl_rate:.2f}%')
print(f'  Chi-square statistic:    {chi2:.4f}')
print(f'  p-value:                 {p_value:.4f}')
print(f'  Degrees of freedom:      {dof}')
print()
if p_value > 0.05:
    print('Conclusion: NOT statistically significant (p > 0.05)')
    print('  This is a monitoring hypothesis, not a confirmed root cause.')
    print('  The 49% weather-coverage gap must be closed before this')
    print('  threshold can be promoted to an actionable SOP.')
else:
    print('Conclusion: STATISTICALLY SIGNIFICANT (p <= 0.05)')
print('=================================================================')
```
